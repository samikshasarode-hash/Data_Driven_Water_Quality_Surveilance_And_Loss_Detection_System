```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside the notebook
%matplotlib inline

# Set plot size
plt.rcParams["figure.figsize"] = (10,6)
```

```python
#load the excel dataset
df = pd.read_excel("../../datasets/147faa26-b07a-4ead-a184-e03e3824f944.xlsx")
```

```python
#display the first five rows
df.head()
```

```python
#view the last five records
df.tail()
```

```python
#check the dataset shape
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
#display column names
df.columns
```

```python
#check the data types
df.dtypes
```

```python
#dataset information
df.info()
```

```python
#check missing values
df.isnull().sum()
```

```python
#calculate missing value percentage
(df.isnull().sum()/len(df))*100
```

```python
#check duplicate records
df.duplicated().sum()
```

```python
#remove duplicate records
df = df.drop_duplicates()
```

```python
#statistical summary
df.describe()
```

```python
#unique values
df.nunique()
```

```python
#correlation matrix
corr = df.corr(numeric_only=True)
corr
```

```python
#correlation heatmap
plt.figure(figsize=(14,10))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
#histogram of all numerical columns
df.hist(figsize=(18,15),
        bins=20)

plt.show()
```

```python
#detect outliers
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    plt.figure(figsize=(6,4))
    sns.boxplot(y=df[col])
    plt.title(col)
    plt.show()
```

```python
#distribution plot
for col in numeric_columns:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
#Check the relationship between TDS and WQI.
sns.scatterplot(
    x="tds_ppm",
    y="wqi_score",
    hue="water_quality_category",
    data=df
)
plt.show()
```

```python
#pair plot
sns.pairplot(df[
[
'water_temperature_c',
'tds_ppm',
'turbidity_ntu',
'wqi_score'
]
])
plt.show()
```

```python
#count plot
plt.figure(figsize=(8,5))

sns.countplot(
    x="water_quality_category",
    data=df
)
plt.xticks(rotation=45)
plt.show()
```

```python
#leakage count
sns.countplot(
    x="leakage_detected",
    data=df
)

plt.show()
```

```python
#alert level count
sns.countplot(
    x="alert_level",
    data=df
)

plt.show()
```

```python
#device status count
sns.countplot(
    x="device_status",
    data=df
)

plt.show()
```

```python
#pie chart
df["water_quality_category"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.show()
```

```python
#skewness
df.skew(numeric_only=True)
```

```python
#kurtosis
df.kurt(numeric_only=True)
```

```python
#covariance matrix
df.cov(numeric_only=True)
```

```python
#final EDA summary
print("========== EDA REPORT ==========")

print("Shape :", df.shape)

print("Missing Values :")
print(df.isnull().sum())

print()

print("Duplicate Rows :", df.duplicated().sum())

print()

print("Statistical Summary")
print(df.describe())
```
